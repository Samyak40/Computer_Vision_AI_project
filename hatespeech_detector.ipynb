```python
import pandas as pd
import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score, precision_score, recall_score, accuracy_score

# 1. Load Datasets
train_df = pd.read_csv('train-3.csv')
test_df = pd.read_csv('test-3.csv')
sample_sub = pd.read_csv('sample_submission(1).csv')

X = train_df['text'].values
y = train_df['label'].values

# 2. Cross-Validation Benchmark
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
f1_scores = []

for train_idx, val_idx in skf.split(X, y):
    X_tr, X_val = X[train_idx], X[val_idx]
    y_tr, y_val = y[train_idx], y[val_idx]

    vec = TfidfVectorizer(ngram_range=(1, 3))
    X_tr_vec = vec.fit_transform(X_tr)
    X_val_vec = vec.transform(X_val)

    clf = LogisticRegression(random_state=42)
    clf.fit(X_tr_vec, y_tr)
    preds = clf.predict(X_val_vec)
    f1_scores.append(f1_score(y_val, preds))

print(f"5-Fold Mean Cross-Validation Binary F1 Score: {np.mean(f1_scores):.4f}")

# 3. Final Model Training on Full Data
full_vectorizer = TfidfVectorizer(ngram_range=(1, 3))
X_train_full = full_vectorizer.fit_transform(X)
X_test_full = full_vectorizer.transform(test_df['text'])

final_model = LogisticRegression(random_state=42)
final_model.fit(X_train_full, y)

# 4. Generate Test Predictions
test_df['label'] = final_model.predict(X_test_full)

# 5. Format Submission
submission = test_df[['id', 'label']]
submission.to_csv('submission.csv', index=False)

print("\nSubmission generated successfully!")
print(submission.head())
```
